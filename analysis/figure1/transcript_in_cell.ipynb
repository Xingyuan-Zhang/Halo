```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# Data as a Series
s = pd.Series(
    {
        'glioblastoma': 90.6,
        'breast cancer': 89.3,
        'lung adenocarcinoma': 82.0,
        'ovarian cancer': 88.4,
        'melanoma': 83.2,
        'pancreatic cancer': 89.1,
        'cervical cancer': 86.7,
        'prostate cancer': 73.7,
        'lymph node': 94.5,
        'mouse brain coronal': 58.0,
        'mouse colon': 82.8,
        'mouse pup': 82.8
    },
    name="percent_within_cells"
)

df = s.rename_axis("dataset").reset_index()

from matplotlib import colors as mcolors

# Sort so the figure reads cleanly (low -> high)
df = df.sort_values("percent_within_cells", ascending=True)
values = df["percent_within_cells"].to_numpy()
# Higher values => less transparent (higher alpha)
alphas = np.interp(values, (values.min(), values.max()), (0.30, 0.90))
base = mcolors.to_rgba("#F28E2B")
bar_colors = [(base[0], base[1], base[2], a) for a in alphas]

# --- publication-style defaults ---
plt.rcParams.update({
    "figure.dpi": 150,
    "savefig.dpi": 600,
    "font.size": 14,
    "axes.titlesize": 13,
    "axes.labelsize": 14,
    "axes.linewidth": 1.0,
    "xtick.labelsize": 12,
    "ytick.labelsize": 14,
    "xtick.major.size": 4,
    "ytick.major.size": 4,
    "xtick.major.width": 1.0,
    "ytick.major.width": 1.0,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "font.sans-serif": ["Arial", "Helvetica", "DejaVu Sans"],
})

# --- figure ---
fig, ax = plt.subplots(figsize=(6.4, 4.6))
ax.barh(
    df["dataset"],
    df["percent_within_cells"],
    color=bar_colors,
    height=0.75,
    edgecolor="none"
)

ax.set_xlabel("Percent transcripts within cells")
ax.set_ylabel("")

# axis styling
ax.set_xlim(0, 100)
ax.set_xticks([0, 20, 40, 60, 80, 100])
ax.xaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x:.0f}%"))

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_linewidth(1.0)
ax.spines["bottom"].set_linewidth(1.0)

ax.margins(y=0.01)

plt.tight_layout(pad=0.3)
plt.show()
```
